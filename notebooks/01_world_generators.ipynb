## Statistical Validation: World A vs World B

Generate both worlds and compare their observable statistics - P(A), P(B), P(A,B), correlation, and fraction where A equals B. The goal is for these to be as close as possible, so that any later difference in SAE 
representations can't simply be attributed to different input statistics.

```python
import sys
sys.path.append('..')

import numpy as np
import pandas as pd
from src.generators import generate_world_a, generate_world_b

def compute_stats(X, label):
    A = X[:, 0]
    B = X[:, 1]
    return {
        "world": label,
        "P(A)": A.mean(),
        "P(B)": B.mean(),
        "P(A,B)": (A & B).mean(),
        "corr(A,B)": np.corrcoef(A, B)[0, 1],
        "frac_equal": (A == B).mean()
    }

X = generate_world_a(n_samples=1000, seed=42)
X_b = generate_world_b(n_samples=1000, seed=42)

stats_a = compute_stats(X, "World A")
stats_b = compute_stats(X_b, "World B")

df = pd.DataFrame([stats_a, stats_b])
print(df)
```

## Parameter Sweep: Matching Correlation

World B's initial correlation (with p=0.9) undershot World A's target correlation (0.779). This sweep tests several values of p to find the one that brings World B's correlation closest to World A's, while keeping the 
causal structure (common cause C) unchanged.

```python
target_corr = stats_a["corr(A,B)"]  # World A's correlation, our target

for p in [0.90, 0.92, 0.94, 0.95, 0.96, 0.98]:
    X_test = generate_world_b(n_samples=1000, seed=42,
                                p_a_given_c1=p, p_a_given_c0=1-p,
                                p_b_given_c1=p, p_b_given_c0=1-p)
    stats_test = compute_stats(X_test, f"p={p}")
    print(f"p={p}: corr(A,B)={stats_test['corr(A,B)']:.3f}  (target={target_corr:.3f})")
```

## Robustness Check: Multiple Seeds

Confirming the statistical match (using p=0.94) holds across different random seeds, not just the one seed used during tuning. A match based on a single lucky seed wouldn't be trustworthy.

```python
for seed in [1, 2, 3, 4, 5]:
    Xa = generate_world_a(n_samples=1000, seed=seed)
    Xb = generate_world_b(n_samples=1000, seed=seed)
    sa = compute_stats(Xa, "A")
    sb = compute_stats(Xb, "B")
    print(f"seed={seed}: corr_A={sa['corr(A,B)']:.3f}  corr_B={sb['corr(A,B)']:.3f}  diff={abs(sa['corr(A,B)']-sb['corr(A,B)']):.3f}")
```
